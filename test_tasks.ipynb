```python
import pandas as pd
from scipy import stats
```

```python
df = pd.read_csv('Данные для тестового задания - Данные АБ тестов.csv')

display(df.head())
print(df['experiment_num'].unique())
print(df['experiment_group'].unique())
```

```python
for exp, sub in df.groupby('experiment_num'):
    control = sub.loc[sub['experiment_group']=='control', 'revenue'].fillna(0).values
    test    = sub.loc[sub['experiment_group']=='test',    'revenue'].fillna(0).values

    arpu_c = control.mean()
    arpu_t = test.mean()
    lift_pct = (arpu_t - arpu_c) / arpu_c * 100 if arpu_c != 0 else np.nan

    t_stat, p_val = stats.ttest_ind(test, control, equal_var=False, nan_policy='omit')

    print(f"--- Эксперимент {int(exp)} ---")
    print(f"n_control={len(control)}  n_test={len(test)}")
    print(f"ARPU control={arpu_c:.2f}  ARPU test={arpu_t:.2f}  lift={lift_pct:.2f}%  p={p_val:.4f}")

    if p_val < 0.05:
        if lift_pct > 0:
            print("Вывод: различия статистически значимы. Тестовая группа показывает более высокий ARPU.")
            print("Рекомендация: внедрять тестовый вариант.\n")
        else:
            print("Вывод: различия статистически значимы. Контрольная группа показывает более высокий ARPU.")
            print("Рекомендация: оставить контрольный вариант.\n")
    else:
         print("Вывод: различия статистически незначимы (p >= 0.05).")
         print("Рекомендация: оставить контрольный вариант или продолжить тест для набора данных.\n")
```
